```python
def run_model(image):
    return image
```

```python
# Build Gradio UI
interface = gr.Interface(
    fn=run_model,
    inputs=gr.Image(type='pil', label='Upload Image'),
    outputs=gr.Image(label='Output'),
    title='Gradio Test',
)
interface.launch(debug=True, share=True)
```
